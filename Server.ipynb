```python
# =========================
# Prepare Server Source
# =========================

import os
import pandas as pd
from google.colab import drive

# 1. Mount Google Drive
drive.mount("/content/drive")

# 2. Correct VFL folder path
server_dir = "/content/drive/MyDrive/VFL_dataset"

# 3. Check files exist
print(os.listdir(server_dir))

# 4. Load server CSV files
server_dir = "/content/drive/MyDrive/milk10k/VFL_dataset"

server_train = pd.read_csv(os.path.join(server_dir, "server_train.csv"))
server_val   = pd.read_csv(os.path.join(server_dir, "server_val.csv"))
server_test  = pd.read_csv(os.path.join(server_dir, "server_test.csv"))

# 5. Clean column names
server_train.columns = server_train.columns.str.strip()
server_val.columns   = server_val.columns.str.strip()
server_test.columns  = server_test.columns.str.strip()

# 6. Check data
print("Train:", server_train.shape)
print("Val:", server_val.shape)
print("Test:", server_test.shape)

print(server_train.head())

# 7. Label mapping
label_map = {
    "benign": 0,
    "low": 1,
    "medium_High_risk": 2
}

# 8. Add numeric labels
server_train["label"] = server_train["risk_group"].map(label_map)
server_val["label"]   = server_val["risk_group"].map(label_map)
server_test["label"]  = server_test["risk_group"].map(label_map)

# 9. Check labels
print("\nTrain labels:")
print(server_train["risk_group"].value_counts())

print("\nVal labels:")
print(server_val["risk_group"].value_counts())

print("\nTest labels:")
print(server_test["risk_group"].value_counts())

# 10. Create fast label lookup dictionaries
train_label_dict = dict(zip(server_train["isic_id"], server_train["label"]))
val_label_dict   = dict(zip(server_val["isic_id"], server_val["label"]))
test_label_dict  = dict(zip(server_test["isic_id"], server_test["label"]))

# 11. Final check
print("\nServer ready!")
print("Train label dict size:", len(train_label_dict))
print("Val label dict size:", len(val_label_dict))
print("Test label dict size:", len(test_label_dict))
```
